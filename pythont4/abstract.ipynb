```python
from abc import ABC,abstractmethod
class Defence(ABC):#abstract class requirement must inherit ABC class and should have 1 abstract method
    @abstractmethod
    def gun(self):
        print("helo")
class Army(Defence): #object can be created even if it doest not implement the method
    pass
class navy(Army):#finally method is implemented else will give error
    def gun(self):
        print("ak47")

d=navy()
d.gun()
    
```
