```python
class Book:
    def __init__(self,bname,date,isbn):
        self.bname=bname
        self.date=date
        self.isbn=isbn
    def show(self):
        print(self.bname)
        print(self.date)
        print(self.isbn) 
class Author(Book):
    def __init__(self,bname,isbn,date,author_name):
        self.aname=author_name
        super().__init__(bname,date,isbn)
        self.show()
    def show(self):
        print(self.aname)
        super().show()
a=Author("bd","123456789","24-02-2026","Murakami")        
```
