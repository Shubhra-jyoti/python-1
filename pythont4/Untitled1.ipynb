```python
import matplotlib.pyplot as plt
phase=["T1","T2","T3","T4"]
marks=[20,22,18.5,17]
plt.subplot(2,3,1)       
plt.plot(phase,marks,marker="v",ms=20,mec="yellow",mfc="green",color="blue",linestyle=":",linewidth=5)
f1={"family":"sans-serif","color":"red","fontsize":15}
f2={"family":"serif","color":"blue","fontsize":20}
f3={"family":"serif","color":"green","fontsize":25}
plt.xlabel("Phase",fontdict=f1)
plt.ylabel("Marks",fontdict=f2)
plt.title("Phase vs Marks",fontdict=f3)
plt.subplot(2,3,2)
plt.bar(phase,marks,color="red",width=0.5)
plt.subplot(2,3,3)
plt.barh(phase,marks,color="blue")
plt.subplot(2,3,4)
plt.pie(marks,labels=phase)
plt.legend()

plt.show()
```
